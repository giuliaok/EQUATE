```python
overall_index = pd.read_csv('/Users/go202/Downloads/complete_index.csv')
df = pd.read_csv('/Users/go202/Downloads/huggingface_languages_timeseries.csv')

import matplotlib.pyplot as plt
import pandas as pd

# Ensure 'date' is a datetime object and 'models' is numeric
df['date'] = pd.to_datetime(df['date'])
df['models'] = pd.to_numeric(df['models'])

# --- Part 1: Calculate and Plot Number of Languages with >= 100 Models per Date ---

# 1. Group by date and language, and sum the 'models' for each group
language_models_per_date = df.groupby(['date', 'code'])['models'].sum().reset_index()

# 2. Filter for languages with at least 100 models
languages_with_100_plus_models_filtered = language_models_per_date[language_models_per_date['models'] >= 100]

# 3. Count unique languages per date that meet the criteria
languages_per_month_100_plus_count = languages_with_100_plus_models_filtered.groupby('date')['code'].nunique().reset_index(name="num_languages_100_plus")
languages_per_month_100_plus_count = languages_per_month_100_plus_count[
    languages_per_month_100_plus_count["num_languages_100_plus"] <= 300
]

languages_by_date_names = {}
for date, group in languages_with_100_plus_models_filtered.groupby('date'):
    languages_by_date_names[date.strftime('%Y-%m')] = group['code'].tolist()
all_name_mappings = {**iso_3_map, **iso3_extra_map}

languages_by_date_names_mapped = {}
for date, group in languages_with_100_plus_models_filtered.groupby('date'):
    original_lang_list = group['code'].tolist()
    mapped_lang_list = []
    for lang_name in original_lang_list:
        # Apply mapping: try iso3map, then iso_extra_map, otherwise keep original
        # The .get() method allows a default value (original name) if key not found
        mapped_name = all_name_mappings.get(lang_name, lang_name)
        mapped_lang_list.append(mapped_name)
    
    languages_by_date_names_mapped[date.strftime('%Y-%m')] = mapped_lang_list
codes_speakers = overall_index[['iso_code', 'n_speakers']]
speakers_by_date = {}

# Iterate through the languages_by_date_names_mapped dictionary
for date, iso_codes in languages_by_date_names_mapped.items():
    # Filter the DataFrame for the languages present on the current date
    filtered_df = codes_speakers[codes_speakers['iso_code'].isin(iso_codes)]

    # Sum the 'n_speakers' for the filtered languages
    total_speakers = filtered_df['n_speakers'].sum()

    # Store the result in the speakers_by_date dictionary
    speakers_by_date[date] = total_speakers

# Print the results
for date, total_speakers in speakers_by_date.items():
    print(f"Date: {date}, Total Speakers: {total_speakers}")
percentage_covered_by_date = {}
all_speakers = codes_speakers['n_speakers'].sum()
# Calculate the percentage for each date
for date, total_speakers in speakers_by_date.items():
    percentage = (total_speakers / all_speakers) * 100
    percentage_covered_by_date[date] = percentage
```

```python
def linearize_gompertz(value, max_value, epsilon=1e-9):
    """
    Applies the Gompertz linearization transformation.
    Handles values close to 0 or max_value by adding a small epsilon and clipping.
    """
    normalized_value = value / max_value
    normalized_value_clipped = np.clip(normalized_value, epsilon, 1 - epsilon)
    return -np.log(-np.log(normalized_value_clipped))

def plot_linearized_gompertz(dataframes_dict, title="Linearized Gompertz Curves - Normalized", y_axis_max_limit=5):
    """
    Replicates the linearized Gompertz plot for multiple technologies.

    Args:
        dataframes_dict (dict): A dictionary where keys are technology names (str)
                                and values are pandas DataFrames with 'Year' and 'Value' columns.
        title (str): The title of the plot.
        y_axis_max_limit (int/float): The maximum value for the y-axis (Linearized Gompertz Curve).
                                      Any points above this will be removed.
    """
    plt.figure(figsize=(10, 7))

    x_ideal = np.linspace(-3, y_axis_max_limit, 100)
    plt.plot(x_ideal, x_ideal, 'k--', label='Ideal Gompertz (Slope 1)', alpha=0.7)

    plot_data = []

    for tech_name, df in dataframes_dict.items():
        if not isinstance(df, pd.DataFrame) or 'Year' not in df.columns or 'Value' not in df.columns:
            print(f"Skipping '{tech_name}': DataFrame must be a pandas DataFrame and contain 'Year' and 'Value' columns.")
            continue

        df_copy = df.copy()

        # --- IMPORTANT CHANGE: Filter out rows where Value is 0 for 'Mobile Phones' ---
        # Note: In the user's snippet, it was 'World Phones', but the dictionary uses 'Mobile Phones'.
        # I'm applying this specific filter only to 'Mobile Phones' as it's the key used in the dict.
        if tech_name == 'Mobile Phones':
            initial_rows = len(df_copy)
            df_copy = df_copy[df_copy['Value'] > 0]
            if len(df_copy) < initial_rows:
                print(f"Removed {initial_rows - len(df_copy)} rows with Value = 0 from '{tech_name}'.")
            if df_copy.empty:
                print(f"Warning: '{tech_name}' DataFrame is empty after filtering out Value = 0. Skipping.")
                continue

        # --- 1. Normalize 'Value' and apply Gompertz linearization ---
        max_val = df_copy['Value'].max()
        if max_val == 0:
            print(f"Warning: Max value for '{tech_name}' is 0 after filtering. Cannot perform Gompertz linearization. Skipping.")
            continue

        df_copy['linearized_gompertz_value'] = df_copy['Value'].apply(
            lambda x: linearize_gompertz(x, max_val)
        )

        # --- 2. Normalize 'Year' (Time - Normalized) using Z-score normalization ---
        if df_copy['Year'].std() == 0:
            print(f"Warning: Standard deviation of Years for '{tech_name}' is 0. Cannot normalize time.")
            df_copy['normalized_time'] = 0
        else:
            df_copy['normalized_time'] = (df_copy['Year'] - df_copy['Year'].mean()) / df_copy['Year'].std()

        df_copy['Technology'] = tech_name
        plot_data.append(df_copy)

    if not plot_data:
        print("No valid dataframes were processed for plotting. Please check your input dataframes.")
        plt.close()
        return

    combined_df = pd.concat(plot_data)

    # Filter out points exceeding y_axis_max_limit after linearization
    combined_df_filtered = combined_df[combined_df['linearized_gompertz_value'] <= y_axis_max_limit]

    # Define a colorblind-friendly palette
    num_technologies = len(combined_df_filtered['Technology'].unique())
    colors = sns.color_palette("colorblind", n_colors=num_technologies)
    markers = ['o', 's', '^', 'D', 'X', 'P', 'v', '<', '>'] # Add more if needed
    linestyles = ['-', '--', ':', '-.', (0, (3, 1, 1, 1)), (0, (5, 1))] # Add more if needed

    for i, tech_name in enumerate(combined_df_filtered['Technology'].unique()):
        tech_df = combined_df_filtered[combined_df_filtered['Technology'] == tech_name]

        # --- MODIFICATION: Adjust alpha based on tech_name ---
        current_alpha = 0.8 # Default prominent alpha
        if tech_name != 'Language Models': # Apply transparency to all except Language Models
            current_alpha = 0.3 # Make it more transparent

        sns.lineplot(
            data=tech_df,
            x='normalized_time',
            y='linearized_gompertz_value',
            label=tech_name,
            color=colors[i % len(colors)],
            marker=markers[i % len(markers)],
            linestyle=linestyles[i % len(linestyles)],
            linewidth=2,
            markersize=6,
            alpha=current_alpha # Use the adjusted alpha
        )

    plt.xlabel("Time - Normalized")
    plt.ylabel("Linearized Gompertz Curve - Normalized")
    plt.title(title)
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.legend(title="Technology", bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.ylim(bottom=None, top=y_axis_max_limit)
    plt.savefig('gomp_colorblind_nlp_2.pdf')
    plt.show()

# --- Example Dummy Data (Replace with your actual data if available) ---

# --- Pre-processing for all DataFrames ---
# Filter out rows where 'Value' is 0 for 'World Phones' (if applicable)
world_phones = world_phones[world_phones['Value'] != 0]

# Convert 'Month' to a single 'Year' column representing fractional years for world_ev
if 'Month' in world_ev.columns:
    world_ev['Month'] = pd.to_datetime(world_ev['Month'])
    world_ev['Year'] = world_ev['Month'].dt.year + (world_ev['Month'].dt.month - 1) / 12
else:
    print("Warning: 'world_ev' does not have a 'Month' column. Ensure 'Year' column is properly set.")

# Convert 'Month' to a single 'Year' column representing fractional years for nlp_world
if 'Month' in nlp_world.columns:
    nlp_world['Month'] = pd.to_datetime(nlp_world['Month'])
    nlp_world['Year'] = nlp_world['Month'].dt.year + (nlp_world['Month'].dt.month - 1) / 12
else:
    print("Warning: 'nlp_world' does not have a 'Month' column. Ensure 'Year' column is properly set.")


# --- Your DataFrames for Plotting ---
your_dataframes_for_plotting = {
     'Mobile Phones': world_phones,
      'Fixed Broadband': world_internet,
      'Electric Vehicles': world_ev,
      'Language Models': nlp_world, 
    'Personal Computers': world_pcs # Added your new dataset here
}

# Run the function with the updated dictionary
plot_linearized_gompertz(your_dataframes_for_plotting, y_axis_max_limit=5)
```
