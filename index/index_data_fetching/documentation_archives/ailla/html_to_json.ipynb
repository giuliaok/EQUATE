# AILLA — HTML to JSON

Parses the raw AILLA resource pages into structured JSON.

```python
import requests
from bs4 import BeautifulSoup
import json
```

```python
url = 'https://www.ailla.utexas.org/sets/408/'
# url = "https://www.ailla.utexas.org/sets/5002/"
response = requests.get(url)

# Step 2: Parse the HTML content using BeautifulSoup
soup = BeautifulSoup(response.content, 'html.parser')
```

```python
def parse_html_file(file_path):
    with open(file_path, 'r', encoding='utf-8') as file:
        html_content = file.read()
    # Parse the HTML content using BeautifulSoup
    soup = BeautifulSoup(html_content, 'html.parser')  # You can use 'html.parser' as well
    return soup
```

```python
data_dic = {}
```

```python
for i in range(10000):
    
    file_path = './raw/set_' + str(i) + '.html' 
    soup = parse_html_file(file_path)
    content = soup.find_all(type="application/json")[0].contents[0]
    data = json.loads(content)
    data_dic[i] = data
```

```python
all_files = []
```

```python
for set_id, data in data_dic.items():
#     print(set_id)
    if "error" in data["props"]["pageProps"]:
        if not data["props"]["pageProps"]["error"] == "Error fetching item data":
            print(data["props"]["pageProps"])
#         assert data["props"]["pageProps"]["error"] == "Error fetching item data"
        continue
    for this_file in data["props"]["pageProps"]["data"]["files"]:
        all_files.append(this_file)
```

```python
len(all_files)
```

```python
import json

def save_json_to_file(data, file_name):
    """Save a dictionary as a JSON file."""
    with open(file_name, 'w', encoding='utf-8') as file:
        json.dump(data, file, indent=4, ensure_ascii=False)

        
save_json_to_file(all_files, "./all_files.json")
```
