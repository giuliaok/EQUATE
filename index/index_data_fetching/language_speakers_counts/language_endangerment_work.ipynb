# Utils

```python
def find_speakers(language):
    # Construct the Wikipedia URL for the language
    url = f"https://en.wikipedia.org/wiki/{language.replace(' ', '_')}"

    # Fetch the page content
    response = requests.get(url)
    if response.status_code != 200:
        return None  # If page not found, return None

    # Parse the HTML content
    soup = BeautifulSoup(response.text, 'html.parser')

    # Find the infobox
    infobox = soup.find('table', class_='infobox')

    # Initialize speakers variable
    speakers = None

    # Find the row containing 'speakers'
    if infobox:
        rows = infobox.find_all('tr')
        for row in rows:
            if 'speakers' in row.text.lower():
                # Extract the number of speakers
                speakers = row.find('td').text.strip()
                break

    return speakers
```

```python
def number_extraction_from_info_box(wikipedia_links):
    language_links = [language for language in wikipedia_links if ' language' in language]
    counts = []
    for language in language_links:
        speakers = find_speakers(language)
        counts.append({'Language': language, 'Speakers': speakers})
    return pd.DataFrame(counts)
```

# Australian Aboriginal

```python
austr_abor_soup = BeautifulSoup(requests.get('https://en.wikipedia.org/wiki/List_of_Australian_Aboriginal_languages').text, 'html.parser')
```

```python
data = []
for tr in austr_abor_soup.find_all('tr'):
    row = {}
    # Find all table data in the row
    tds = tr.find_all('td')
    # If there are at least 4 table data elements in the row, extract their text
    if len(tds) >= 4:
        row['language'] = tds[0].get_text(strip=True) if tds and len(tds) > 0 else ''
        row['alternative names'] = tds[1].get_text(strip=True) if tds and len(tds) > 1 else ''
        row['speakers'] = tds[2].get_text(strip=True) if tds and len(tds) > 2 else ''
        row['status'] = tds[3].get_text(strip=True) if tds and len(tds) > 3 else ''
        data.append(row)
```
