# Bible availability

Determines whether a Bible translation exists for each language, used as a
binary availability feature.

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
def get_languages():
    page = 1
    all_languages = []

    while True:
        url = f"https://www.bible.com/languages?page={page}"
        response = requests.get(url)

        # Check if page exists
        if response.status_code != 200:
            print(f"Reached end of pages at page {page}.")
            break

        # Parse the page with BeautifulSoup
        soup = BeautifulSoup(response.text, 'html.parser')
        languages_section = soup.find('h2', class_="text-text-light dark:text-text-dark font-bold text-19 md:text-23 leading-comfy font-aktiv-grotesk pbe-2")

        # If no additional languages found, break
        if not languages_section:
            print(f"No additional languages found on page {page}.")
            break

        # Extract each language and its code
        language_list = languages_section.find_next('ul').find_all('li')
        for language_item in language_list:
            lang_link = language_item.find('a', href=True)
            if lang_link:
                lang_name_code = lang_link.text.strip()
                name, code = lang_name_code.rsplit(' ', 1)
                code = code.strip('()')  # Remove parentheses from code
                all_languages.append({'name': name.strip(), 'code': code})

        print(f"Page {page} processed.")
        page += 1  # Move to the next page

    return all_languages

# Fetch languages and display results
languages = get_languages()
for language in languages:
    print(language)
```

```python
bibles_languages = pd.DataFrame(languages)
```
