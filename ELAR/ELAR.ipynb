```python
def fetch_urls(base_url, num_pages):
    urls = []
    for i in range(1, num_pages + 1):
        url = f"{base_url}?pg={i}"
        response = requests.get(url)
        if response.status_code == 200:
            soup = BeautifulSoup(response.content, 'html.parser')
            divs = soup.find_all('div', class_='one_sixth grey_block new-secondary-background result-item')
            for div in divs:
                a_tag = div.find('a', class_='new-primary new-primary-tint-hover multiline')
                if a_tag:
                    title = a_tag.text.strip()
                    print(title)
                    href = a_tag['href']
                    print(href)
                    urls.append((title, href))

        else:
            print(f"Failed to fetch {url}. Status code: {response.status_code}")
    return urls

base_url = "https://www.elararchive.org/uncategorized/SO_5f038640-311d-4296-a3e9-502e8a18f5b7/"
num_pages = 27

result_urls = fetch_urls(base_url, num_pages)
print(result_urls)
```

```python
df = pd.DataFrame(result_urls, columns = ['collection', 'url_collection'])
```

```python
def fetch_urls_access(base_url):
    urls = []
    page_number = 1
    while True:
        url = f"{base_url}?pg={page_number}"
        response = requests.get(url)
        if response.status_code == 200:
            soup = BeautifulSoup(response.content, 'html.parser')
            divs = soup.find_all('div', class_='one_sixth grey_block new-secondary-background result-item')
            if not divs:
                break
            for div in divs:
                a_tag = div.find('a', class_='new-primary new-primary-tint-hover multiline')
                access_level_tag = div.find('div', class_='archive_data_content col2')
                if a_tag and access_level_tag:
                    title = a_tag.text.strip()
                    print(title)
                    href = a_tag['href']
                    access_level = access_level_tag.find('p').text.strip()
                    urls.append((title, href, access_level))
            page_number += 1
        else:
            print(f"Failed to fetch {url}. Status code: {response.status_code}")
            break
    return urls
```

```python
small_df = df[0:10]
```

```python
small_df['inner_urls'] = small_df['url_collection'].apply(fetch_urls_access)
```
